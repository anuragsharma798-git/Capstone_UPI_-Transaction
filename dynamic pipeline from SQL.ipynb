```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine
```

### **Create Secure Database Connection**

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="Anurag@$123",
    host="localhost",
    port=3306,
    database="capstone"
)

engine = create_engine(url)
```

### **Executing SQL queries from Python using pandas.read_sql_query**

```python
query = "SELECT * FROM customer_master LIMIT 5"

customer_df = pd.read_sql_query(query, engine)
print(customer_df.head())
```

### **Loading data from all tables into Pandas dataframes for further use**

```python
customer_master = pd.read_sql_query(
    "SELECT * FROM customer_master",
    engine
)

merchant_info = pd.read_sql_query(
    "SELECT * FROM merchant_info",
    engine
)

device_info = pd.read_sql_query(
    "SELECT * FROM device_info",
    engine
)

upi_account_details = pd.read_sql_query(
    "SELECT * FROM upi_account_details",
    engine
)

upi_transaction_history = pd.read_sql_query(
    "SELECT * FROM upi_transaction_history",
    engine
)

fraud_alert_history = pd.read_sql_query(
    "SELECT * FROM fraud_alert_history",
    engine
)

customer_feedback_surveys = pd.read_sql_query(
    "SELECT * FROM customer_feedback_surveys",
    engine
)


```
